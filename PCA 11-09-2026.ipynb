# principle component analysis

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer 
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
df=load_breast_cancer()
```

```python
df
```

```python
x=pd.DataFrame(data=df.data,columns=df.feature_names)
```

```python
x
```

```python
y=pd.DataFrame(data=df.target)
```

```python
y
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaled=MinMaxScaler()
```

```python
Scaled_x=scaled.fit_transform(x)
```

```python
x_updated=pd.DataFrame(data=Scaled_x)
```

```python
from sklearn.decomposition import PCA
```

```python
pca_model=PCA(n_components=10)
```

```python
x_final=pca_model.fit_transform(x_updated)
```

```python
pca_model.explained_variance_ratio_
```

```python
import numpy as np
```

```python
np.cumsum(pca_model.explained_variance_ratio_)
```

```python
final_x=pd.DataFrame(data=x_final)
```

```python
x_train,x_test,y_train,y_test=train_test_split(final_x,y,stratify=y,random_state=32,test_size=0.20)
model=DecisionTreeClassifier()
model.fit(x_train,y_train)
y_train_pred=model.predict(x_train)
print(accuracy_score(y_train,y_train_pred))
y_test_pred=model.predict(x_test)
print(accuracy_score(y_test,y_test_pred))
```
